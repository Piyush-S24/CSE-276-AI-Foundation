# 2. Problem Statement

Imagine a campus where an emergency vehicle or student needs to travel from one location to another.

We will model the campus as a **graph**:

- **Node (Vertex):** A location, such as Main Gate or Library.
- **Edge:** A connection or road between two locations.

Our task is:

> **Find a route from the Main Gate to the Library.**


# 3. What is a Graph?

A graph can be represented as:

```text
        Library
       /       \
   Hostel ---- Canteen
      |          |
   Main Gate -- Admin
        \       /
           Lab
```

Each location is a **node**. Each road is an **edge**.

We will store this graph in Python using an **adjacency list**.


````python
campus = {
    "Main Gate": ["Hostel", "Admin", "Lab"],
    "Hostel": ["Main Gate", "Library", "Canteen"],
    "Library": ["Hostel", "Canteen"],
    "Canteen": ["Hostel", "Library", "Admin"],
    "Admin": ["Main Gate", "Canteen", "Lab"],
    "Lab": ["Main Gate", "Admin"]
}

print("Campus Graph:")
for location, connections in campus.items():
    print(f"{location} -> {connections}")
````

# 4. What is a Search Algorithm?

A search algorithm explores possible locations to find a goal.

For example:

- **Start:** Main Gate
- **Goal:** Library

The algorithm must decide:

> Which location should I explore next?

This decision is the main difference between BFS and DFS.


# 5. Queue Before BFS

A **Queue** works like people standing in a line.

```text
FRONT                          BACK
  ↓                              ↓
[A]  [B]  [C]  [D]
```

The first person to enter leaves first.

## FIFO

**First In, First Out**

In Python, BFS commonly uses `deque`:

- `append()` adds an item at the back.
- `popleft()` removes the item at the front.


````python
from collections import deque

queue = deque()

queue.append("Main Gate")
queue.append("Hostel")
queue.append("Admin")

print("Queue after adding locations:", list(queue))

removed = queue.popleft()
print("Removed from front:", removed)
print("Queue now:", list(queue))
````

# 6. Breadth First Search (BFS)

BFS explores the graph **level by level**.

Starting from Main Gate:

```text
Level 0: Main Gate

Level 1: Hostel, Admin, Lab

Level 2: Library, Canteen
```

Because BFS explores the nearest levels first, it finds a path with the **minimum number of edges** in an unweighted graph.

## Why does BFS use a Queue?

The first location discovered is explored first.

Therefore:

**BFS = Queue = FIFO**


## BFS Algorithm

1. Put the starting path into the queue.
2. Remove the first path from the queue.
3. Check the current location.
4. If it is the goal, return the path.
5. Otherwise, add all unvisited neighbouring locations to the queue.
6. Repeat until the goal is found.


````python
def bfs(graph, start, goal):
    # Queue stores complete paths
    queue = deque([[start]])

    # Keeps track of visited locations
    visited = set()

    # Stores the order in which states are explored
    explored = []

    while queue:
        # FIFO: remove the oldest path
        path = queue.popleft()
        current = path[-1]

        # Skip a location that has already been processed
        if current in visited:
            continue

        # Mark current location as visited
        visited.add(current)
        explored.append(current)

        print(f"Exploring: {current}")
        print(f"Current path: {path}")

        # Goal test
        if current == goal:
            return path, explored

        # Add neighbouring paths to the queue
        for neighbour in graph[current]:
            if neighbour not in visited:
                new_path = path + [neighbour]
                queue.append(new_path)

    return None, explored
````

# 7. Run BFS

We will now find a route from **Main Gate** to **Library**.


````python
start = "Main Gate"
goal = "Library"

print("===== BFS EXECUTION =====\n")

bfs_path, bfs_explored = bfs(campus, start, goal)

print("\n===== BFS RESULT =====")
print("Path found:", bfs_path)
print("Path length (number of roads):", len(bfs_path) - 1)
print("Explored states:", bfs_explored)
print("Number of explored states:", len(bfs_explored))
````

# 8. Stack Before DFS

A **Stack** works like a stack of plates.

```text
        TOP
         ↓
       [ D ]   <- Last added
       [ C ]
       [ B ]
       [ A ]   <- First added
```

The last item added is removed first.

## LIFO

**Last In, First Out**

In Python:

- `append()` adds an item to the top.
- `pop()` removes the item from the top.


````python
stack = []

stack.append("Main Gate")
stack.append("Hostel")
stack.append("Library")

print("Stack:", stack)

removed = stack.pop()
print("Removed from top:", removed)
print("Stack now:", stack)
````

# 9. Depth First Search (DFS)

DFS explores one route **as deeply as possible** before trying another route.

For example:

```text
Main Gate
    ↓
 Hostel
    ↓
 Canteen
    ↓
 Admin
```

If the route does not lead to the goal, DFS goes back and tries another branch. This is called **backtracking**.

## Why does DFS use a Stack?

The most recently discovered location is explored first.

Therefore:

**DFS = Stack = LIFO**

> Important: DFS can find a path, but it does **not guarantee the shortest path**.


## DFS Algorithm

1. Put the starting path into the stack.
2. Remove the last path from the stack.
3. Check the current location.
4. If it is the goal, return the path.
5. Otherwise, add unvisited neighbouring locations to the stack.
6. Repeat until the goal is found.


````python
def dfs(graph, start, goal):
    # Stack stores complete paths
    stack = [[start]]

    # Keeps track of visited locations
    visited = set()

    # Stores the order in which states are explored
    explored = []

    while stack:
        # LIFO: remove the newest path
        path = stack.pop()
        current = path[-1]

        # Skip a location that has already been processed
        if current in visited:
            continue

        # Mark current location as visited
        visited.add(current)
        explored.append(current)

        print(f"Exploring: {current}")
        print(f"Current path: {path}")

        # Goal test
        if current == goal:
            return path, explored

        # Add neighbouring paths to the stack
        for neighbour in graph[current]:
            if neighbour not in visited:
                new_path = path + [neighbour]
                stack.append(new_path)

    return None, explored
````

# 10. Run DFS


````python
print("===== DFS EXECUTION =====\n")

dfs_path, dfs_explored = dfs(campus, start, goal)

print("\n===== DFS RESULT =====")
print("Path found:", dfs_path)
print("Path length (number of roads):", len(dfs_path) - 1)
print("Explored states:", dfs_explored)
print("Number of explored states:", len(dfs_explored))
````

# 11. Compare Execution Time

For a fair comparison, we will create quieter versions of BFS and DFS without print statements.

We use `time.perf_counter()` to measure elapsed execution time.

> Because our campus graph is very small, execution times may be extremely tiny and can vary between runs. The comparison demonstrates **how to measure algorithm performance**, not that one algorithm is always faster.


````python
import time

def bfs_quiet(graph, start, goal):
    queue = deque([[start]])
    visited = set()
    explored = []

    while queue:
        path = queue.popleft()
        current = path[-1]

        if current in visited:
            continue

        visited.add(current)
        explored.append(current)

        if current == goal:
            return path, explored

        for neighbour in graph[current]:
            if neighbour not in visited:
                queue.append(path + [neighbour])

    return None, explored


def dfs_quiet(graph, start, goal):
    stack = [[start]]
    visited = set()
    explored = []

    while stack:
        path = stack.pop()
        current = path[-1]

        if current in visited:
            continue

        visited.add(current)
        explored.append(current)

        if current == goal:
            return path, explored

        for neighbour in graph[current]:
            if neighbour not in visited:
                stack.append(path + [neighbour])

    return None, explored
````

````python
# Measure BFS time
start_time = time.perf_counter()
bfs_path, bfs_explored = bfs_quiet(campus, start, goal)
bfs_time = time.perf_counter() - start_time

# Measure DFS time
start_time = time.perf_counter()
dfs_path, dfs_explored = dfs_quiet(campus, start, goal)
dfs_time = time.perf_counter() - start_time

print("BFS Time:", bfs_time, "seconds")
print("DFS Time:", dfs_time, "seconds")
````

# 14. BFS vs DFS — Conceptual Comparison

| Feature | BFS | DFS |
|---|---|---|
| Full Name | Breadth First Search | Depth First Search |
| Main Strategy | Explore level by level | Explore deeply first |
| Main Data Structure | Queue | Stack |
| Order | FIFO | LIFO |
| Shortest Path in Unweighted Graph | Yes | Not guaranteed |
| Memory Requirement | Usually higher | Often lower |
| Best for | Nearest/shortest-step route | Deep exploration/path existence |
| Backtracking Behaviour | Explores breadth first | Naturally goes deep and backtracks |


# 15. When is BFS Better?

BFS is better when:

1. We need the **shortest path in terms of number of edges**.
2. The destination is likely to be close to the starting point.
3. All roads/connections are treated as having equal cost.

### Example: Campus Emergency Response

Suppose an emergency vehicle needs to reach an incident.

If every road is treated equally, BFS can find the route requiring the fewest road segments.

**Example question:**

> Which route reaches the emergency location using the fewest connections?

**Recommended algorithm: BFS**


# 16. When is DFS Better?

DFS can be useful when:

1. We want to explore a route deeply.
2. We mainly want to know whether a path exists.
3. Memory is limited compared with a wide search.
4. The solution may be deep in the search structure.

### Important Note

DFS may reach the goal through a longer route even when a shorter route exists.

Therefore, DFS is generally **not the preferred algorithm when shortest path quality is the main requirement**.


# 17. Important Limitation for Real Route Planning

This practical uses an **unweighted graph**.

Real navigation systems may need to consider:

- Different road distances
- Traffic
- Travel time
- Road closures
- Fuel cost
- Emergency priority

In these situations, BFS and DFS alone may not be sufficient.

More suitable algorithms include:

- Dijkstra's Algorithm
- A* Search

However, BFS and DFS are important classical search algorithms because they teach the fundamental ideas of **state-space search, exploration strategies, queues, and stacks**.

In this practical, we developed an intelligent route planner using two classical AI search algorithms.

### BFS
- Uses a **Queue**
- Follows **FIFO**
- Explores level by level
- Guarantees the shortest path in an unweighted graph

### DFS
- Uses a **Stack**
- Follows **LIFO**
- Explores one route deeply before trying alternatives

### Final Decision

For an emergency or campus navigation problem where we want the route with the **fewest connections**, **BFS is generally the better choice**.
